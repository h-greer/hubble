```python
wid = 80
oversample = 4

nwavels = 20
npoly=8

n_modes = 45
n_zernikes = 30  # as in the calibrator fits, whose low-order span (Noll 1-33) was removed from the PCA

resolved_wid = 1

klip_file = "../data/wavefront_pca.npz"
n_klip = np.load(klip_file)["modes"].shape[0]
amp_file = "../data/amplitude_pca.npz"
n_amp_klip = np.load(amp_file)["modes"].shape[0]
optics = NICMOSCoronagraph(512, wid, oversample, n_modes=n_modes, n_zernikes=n_zernikes, turboklip=klip_file, amp_klip=amp_file)

detector = NICMOSDetector(oversample, wid)

spectrum_basis = np.ones((nwavels, npoly))


ddir = '../data/NICMOS-LAPL-DD1/archive.stsci.edu/missions/hlsp/laplace/dd1/LAPL/NICMOS-LAPL-DD1/LAPL_DATA/contemp_flats/repaired/'
flatdir = '../data/NICMOS-LAPL-DD1/archive.stsci.edu/missions/hlsp/laplace/dd1/LAPL/NICMOS-LAPL-DD1/HOLEFLATS/'

vects = np.load("../data/iterative_spectrum_basis_F160W.npy")[:,:npoly]
assert vects.shape == (nwavels, npoly)
spectrum_basis = vects/np.sqrt(np.mean(vects**2, axis=0))



# As in batch/calibrator-joint.py: the cold mask lateral position (shift) and its tilt (a pupil-plane Tilt
# initialised from each exposure's own TARSIAFX/Y header, i.e. pointing) can differ between exposures
class PerExposureColdMaskFit(SinglePointFit):
    PARAM_KEYS = SinglePointFit.PARAM_KEYS | {"cold_mask_shift": "exposure", "cold_mask_tilt": "exposure"}

# HR 8799: same visit pair at two rolls (ORIENTAT -147.4 and -117.5 deg); the spectrum is keyed by target, so both share it
exposures_single = [
    exposure_from_file(ddir + f'{root}_clc_calf.fits', PerExposureColdMaskFit(spectrum_basis, "F160W"), crop=wid, extra_bad=None, flatcorr=flatdir)
    for root in ["n4qs09akq", "n4qs10asq"]
]
```

```python
for exp in exposures_single:
    print(exp.target)
```

```python
params = {
    "spectrum": {},
    "primary_klip": {},
    "primary_low": {},
    "primary_tilt": {},
    "cold_mask_tilt": {},
    "cold_mask_shift": {},
    # "jitter": {},
    "bias": {},
    # Keyed globally: amplitude is fixed per run in the calibrator fits, and per-exposure freedom could absorb planets
    "primary_amp_klip": {"global": np.zeros(n_amp_klip)},
}

# Median global optical parameters of the calibrator fits (notebooks/calibrator-pca.ipynb)
optical_params = np.load("../data/calibrator_optical_params.npy", allow_pickle=True)[()]

for idx, exp in enumerate(exposures_single):
    params["spectrum"][exp.fit.get_key(exp, "spectrum")] = (np.zeros(npoly)).at[0].set((np.nansum(exp.data)/nwavels)*100)

    params["primary_tilt"][exp.fit.get_key(exp, "primary_tilt")] = np.array([-0.05, -0.75])*0.075
    params["cold_mask_tilt"][exp.fit.get_key(exp, "cold_mask_tilt")] = np.array([
        np.array(exp.hdr["TARSIAFY"],dtype=float) - (256-44),
        np.array(exp.hdr["TARSIAFX"],dtype=float) - (256-181),
    ])*0.075
    params["cold_mask_shift"][exp.fit.get_key(exp, "cold_mask_shift")] = optical_params["cold_mask_shift"]["global"]

    params["primary_klip"][exp.fit.get_key(exp, "primary_klip")] = np.zeros(n_klip)
    params["primary_low"][exp.fit.get_key(exp, "primary_low")] = np.zeros((n_zernikes))

    params["bias"][exp.fit.get_key(exp, "bias")] = 0.
    # params["jitter"][exp.fit.get_key(exp, "jitter")] = 7. # mas; gradient vanishes at 0

# cold_mask_shift is per exposure here, so the global median only sets its initial value above
params = params | {k: v for k, v in optical_params.items() if k != "cold_mask_shift"}


model_single = set_array(NICMOSModel(exposures_single, params, optics, detector))

params = ModelParams(params)
```
